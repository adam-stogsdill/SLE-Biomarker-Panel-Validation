```python
import pandas as pd
```

```python
pathway_data = pd.read_csv("./data/pathway_data.csv")
pathway_data.head()
```

```python
SLE_genes = []
with open('./data/sle_genes.txt') as sle_file:
    for row in sle_file.readlines():
        gene = row.strip()
        SLE_genes.append(gene)
```

```python
# See if any second degree biomarkers are in the SLE pathway
nodal_connections = []
for i, row in pathway_data.iterrows():
    associated_genes_in_sle = set(eval(row['Associated_Genes'])) & set(SLE_genes)
    if associated_genes_in_sle:
        print(f"Pathway ID: {row['Pathway_ID']}, Desc: {row['Description']}, Gene: {row['gene_name']}")
        print(associated_genes_in_sle)
        print('\n')
```

```python
import pandas as pd
import networkx as nx
from pyvis.network import Network

G = nx.Graph()

unique_gene_names = set()
unique_pathways = set()

for i, row in pathway_data.iterrows():
    associated_genes_in_sle = set(eval(row['Associated_Genes'])) & set(SLE_genes)
    if associated_genes_in_sle:
        #print(f"Pathway ID: {row['Pathway_ID']}, Desc: {row['Description']}, Gene: {row['gene_name']}")
        #print(associated_genes_in_sle)
        #print('\n')

        if row['Description'] not in unique_pathways:
            unique_pathways.add(row['Description'])
            G.add_node(row['Description'], group=1, color='green', title='Pathway')

        for gene in associated_genes_in_sle:
            if gene not in unique_gene_names:
                unique_gene_names.add(gene)
                G.add_node(gene, group=3, color='red', title='SLE GENE')
            G.add_edge(row['Description'], gene)

        if row['gene_name'] not in unique_gene_names:
            unique_gene_names.add(row['gene_name'])
            G.add_node(row['gene_name'], group=2, color='blue', title='Biomarker Gene')
        G.add_edge(row['Description'], row['gene_name'])

# Add physics controls so you can tweak the layout manually in the browser
net = Network(notebook=True, height="750px", width="100%", bgcolor="#FFFFFF", font_color="black")
net.from_nx(G)

net.show_buttons(filter_=['physics']) 
net.show("gene_network.html")  
```
